# Using the Enhanced Twi Text Processor

This notebook demonstrates how to use the enhanced text processing module for Twi/Akan languages.

```python
# Import the processor
# Note: In Colab, rename the uploaded file from '000_enhanced_twi_processor.py' 
# to 'enhanced_twi_processor.py' or adjust the import below
try:
    # Try importing with the numbered prefix (local)
    import importlib.util
    spec = importlib.util.spec_from_file_location("enhanced_twi_processor", "000_enhanced_twi_processor.py")
    enhanced_twi_processor = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(enhanced_twi_processor)
    from enhanced_twi_processor import EnhancedTwiTextProcessor
except:
    # Fallback for when file is renamed in Colab
    from enhanced_twi_processor import EnhancedTwiTextProcessor

# Initialize with your preferred dialect
processor = EnhancedTwiTextProcessor(dialect='asante')
print("Processor initialized for Asante Twi")
```

```python
# Example 1: Process numbers and currency
text1 = "The price is GH₵250 and we need 15 items."
result1 = processor.process_text(text1)

print("Original:", result1['original'])
print("Processed:", result1['processed'])
print()
```

```python
# Example 2: Process dates and times
text2 = "The meeting is on Monday at 10:30 AM."
result2 = processor.process_text(text2)

print("Original:", result2['original'])
print("Processed:", result2['processed'])
print()
```

```python
# Example 3: Process abbreviations
text3 = "Dr. Mensah from KNUST said NPP and NDC are meeting."
result3 = processor.process_text(text3)

print("Original:", result3['original'])
print("Processed:", result3['processed'])
print()
```

```python
# Example 4: Generate phonemes for TTS
text4 = "Akwaaba, ɛte sɛn? Me din de Kofi."
result4 = processor.process_text(text4, generate_phonemes=True)

print("Original:", result4['original'])
print("Processed:", result4['processed'])
print("Phonemes:", result4['phonemes'])
print("Has Twi chars:", result4['has_twi_chars'])
```

```python
# Example 5: Complex mixed text
text5 = """Dr. Akosua will charge $100 or GH₵500 for the consultation 
on Tuesday, December 25, 2024 at 2:30 PM. 
Please call 0244123456 for appointment."""

result5 = processor.process_text(text5, generate_phonemes=False)

print("Original:")
print(result5['original'])
print("\nProcessed:")
print(result5['processed'])
print("\nStats:")
print(f"  Word count: {result5['word_count']}")
print(f"  Char count: {result5['char_count']}")
print(f"  Dialect: {result5['dialect']}")
```

```python
# Example 6: Process with different dialects
text6 = "Dɛn na woreyɛ? Wɔ he na wokɔ?"

# Process as Asante
processor_asante = EnhancedTwiTextProcessor(dialect='asante')
result_asante = processor_asante.process_text(text6)

# Process as Akuapem
processor_akuapem = EnhancedTwiTextProcessor(dialect='akuapem')
result_akuapem = processor_akuapem.process_text(text6)

print("Original:", text6)
print("Asante:", result_asante['processed'])
print("Akuapem:", result_akuapem['processed'])
```

```python
# Example 7: Batch processing (if you have a list of texts)
texts = [
    "The cost is GH₵100",
    "Meet me at 3:00 PM",
    "Today is Monday",
    "Call 0244123456"
]

processed_texts = []
for text in texts:
    result = processor.process_text(text)
    processed_texts.append(result['processed'])
    print(f"Original: {text}")
    print(f"Processed: {result['processed']}")
    print()
```

## Custom Configuration

You can control which processing steps to apply:

```python
text = "Dr. Smith paid $500 on Monday at 10:30"

# Process with all features
result_all = processor.process_text(
    text,
    expand_numbers=True,
    expand_abbreviations=True,
    process_datetime=True,
    apply_dialect=True,
    generate_phonemes=True
)

# Process with only number conversion
result_numbers_only = processor.process_text(
    text,
    expand_numbers=True,
    expand_abbreviations=False,
    process_datetime=False,
    apply_dialect=False,
    generate_phonemes=False
)

print("Full processing:", result_all['processed'])
print("\nNumbers only:", result_numbers_only['processed'])
```
