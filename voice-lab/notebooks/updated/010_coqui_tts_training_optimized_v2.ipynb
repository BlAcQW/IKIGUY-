```python
from typing import Dict, List, Optional, Tuple, Any, Union
from pathlib import Path
import pandas as pd
import soundfile as sf
import numpy as np
from tqdm.auto import tqdm
import re
from collections import defaultdict

# Enhanced path configuration (Moved from environment setup)
import os
import sys

IN_COLAB = 'google.colab' in sys.modules
IN_KAGGLE = 'kaggle_secrets' in sys.modules
IN_LOCAL = not (IN_COLAB or IN_KAGGLE)

if IN_COLAB:
    try:
        from google.colab import drive
        drive.mount('/content/drive')
        BASE_PATH = Path('/content/drive/MyDrive/voice-lab-data')
        print("✅ Google Drive mounted")
    except Exception as e:
        print(f"⚠️ Could not mount Google Drive: {e}")
        BASE_PATH = Path('/content/voice-lab-data')
elif IN_KAGGLE:
    BASE_PATH = Path('/kaggle/working/voice-lab-data')
else:
    BASE_PATH = Path('C:/Projects/kasaai-v2/apps/voice-lab')

# Create comprehensive directory structure
PATHS = {
    'base': BASE_PATH,
    'data': BASE_PATH / 'data',
    'inventory': BASE_PATH / 'data' / 'data_inventory',
    'processed': BASE_PATH / 'data' / 'processed',
    'tts_ready': BASE_PATH / 'data' / 'tts_ready',
    'models': BASE_PATH / 'models',
    'xtts_models': BASE_PATH / 'models' / 'xtts',
    'output': BASE_PATH / 'output',
    'logs': BASE_PATH / 'logs',
    'temp': BASE_PATH / 'temp'
}

for path in PATHS.values():
    path.mkdir(parents=True, exist_ok=True)

print(f"\n📂 Data paths configured:")
print(f"   Base: {PATHS['base']}")
print(f"   Data: {PATHS['data']}")
print(f"   TTS Ready: {PATHS['tts_ready']}")
print(f"   Models: {PATHS['models']}")


class ComprehensiveTTSDataLoader:
    """Load and process comprehensive TTS data inventory with multi-speaker support."""

    def __init__(self, paths: Dict[str, Path]):
        self.paths = paths
        self.tts_inventory = None
        self.speaker_stats = None
        self.dataset_stats = None

        # Enhanced character sets for multi-dialect support
        self.twi_base_chars = "aàáâãäbcdḍeèéêëɛfghiìíîïjklmnñoòóôõöɔpqrstuùúûüvwxyỳýz"
        self.twi_chars_upper = self.twi_base_chars.upper()
        self.punctuation = "!'(),-.:;?¿¡ "
        self.numbers = "0123456789"
        self.special_chars = "@#$%&*+=[]{}/\\|<>~`"

        # Dialect-specific phoneme mappings
        self.dialect_markers = {
            'asante': ['nana', 'otumfuo', 'kumasi'],
            'akuapem': ['akropong', 'mamfe'],
            'fante': ['mfantse', 'cape coast', 'elmina'],
            'ga': ['accra', 'tema']
        }

        # Complete character set
        self.full_charset = list(set(
            self.twi_base_chars + self.twi_chars_upper +
            self.punctuation + self.numbers + self.special_chars
        ))

        print(f"📝 Enhanced character set: {len(self.full_charset)} characters")
        print(f"   Twi characters: {len(set(self.twi_base_chars + self.twi_chars_upper))}")
        print(f"   Full set: {''.join(sorted(self.full_charset)[:50])}...")

    def load_tts_inventory(self) -> pd.DataFrame:
        """Load comprehensive TTS training data inventory."""
        print("📊 Loading comprehensive TTS data inventory...")

        inventory_file = self.paths['inventory'] / 'tts_training_data_inventory.csv'

        if inventory_file.exists():
            try:
                df = pd.read_csv(inventory_file)

                # Clean and process inventory
                df = df.dropna(subset=['raw_data_directory', 'data_name'])

                # Filter out non-applicable entries
                df = df[df['audio_file_directory'] != 'not_applicable']

                print(f"   ✅ Loaded {len(df)} TTS datasets from inventory")

                # Display dataset summary
                print(f"\n📈 Dataset Summary:")
                dataset_counts = df['data_name'].value_counts()
                for dataset, count in dataset_counts.head(10).items():
                    print(f"   {dataset}: {count} entries")

                self.tts_inventory = df
                return df

            except Exception as e:
                print(f"   ⚠️ Error loading TTS inventory: {e}")
                return self._create_minimal_inventory()
        else:
            print(f"   ⚠️ TTS inventory not found at {inventory_file}")
            return self._create_minimal_inventory()

    def _create_minimal_inventory(self) -> pd.DataFrame:
        """Create minimal inventory for testing."""
        print("   Creating minimal test inventory...")

        return pd.DataFrame({
            'raw_data_directory': [],
            'data_name': [],
            'audio_file_directory': [],
            'text_files_directory': [],
            'speaker_embeddings_directory': [],
            'phoneme_alignments_directory': [],
            'prosody_annotations_directory': []
        })

    def discover_audio_text_pairs(self, inventory_row: pd.Series) -> List[Dict]:
        """Discover audio-text pairs from inventory entry."""
        pairs = []

        # Get directories
        audio_dir = Path(inventory_row['audio_file_directory'])
        text_dir = Path(inventory_row['text_files_directory'])
        speaker_dir = Path(inventory_row.get('speaker_embeddings_directory', '')) if pd.notna(inventory_row.get('speaker_embeddings_directory')) else None

        # Check if directories exist
        if not audio_dir.exists() or not text_dir.exists():
            return pairs

        # Find audio files
        audio_extensions = ['.wav', '.mp3', '.flac', '.m4a', '.ogg']
        audio_files = []
        for ext in audio_extensions:
            audio_files.extend(list(audio_dir.glob(f'**/*{ext}')))

        # Find text files
        text_extensions = ['.txt', '.lab', '.csv']
        text_files = []
        for ext in text_extensions:
            text_files.extend(list(text_dir.glob(f'**/*{ext}')))

        # Try to match audio and text files
        for audio_file in audio_files:
            # Look for corresponding text file
            base_name = audio_file.stem

            for text_file in text_files:
                if text_file.stem == base_name or base_name in text_file.stem:
                    try:
                        # Read text content
                        with open(text_file, 'r', encoding='utf-8') as f:
                            text_content = f.read().strip()

                        if text_content:  # Only include non-empty text
                            pair = {
                                'audio_path': str(audio_file),
                                'text': text_content,
                                'dataset': inventory_row['data_name'],
                                'speaker_id': self._extract_speaker_id(audio_file, speaker_dir),
                                'dialect': self._detect_dialect(text_content),
                                'language': self._detect_language(text_content)
                            }
                            pairs.append(pair)
                            break

                    except Exception as e:
                        continue

        return pairs

    def _extract_speaker_id(self, audio_path: Path, speaker_dir: Optional[Path]) -> str:
        """Extract speaker ID from file path or speaker directory."""
        # Try to extract from filename first
        filename = audio_path.stem

        # Common patterns for speaker IDs
        import re

        # Pattern: speaker_001, spk_01, female_001, etc.
        patterns = [
            r'speaker[_-]?(\d+)',
            r'spk[_-]?(\d+)',
            r'(male|female)[_-]?(\d+)',
            r'(\w+)[_-]speaker',
            r'([mf]\d+)',  # m01, f02, etc.
        ]

        for pattern in patterns:
            match = re.search(pattern, filename.lower())
            if match:
                return match.group(0)

        # Extract from parent directory
        parent_parts = audio_path.parts
        for part in reversed(parent_parts):
            if any(keyword in part.lower() for keyword in ['speaker', 'spk', 'voice']):
                return part

        # Default to dataset + filename
        return f"unknown_{filename[:10]}"

    def _detect_dialect(self, text: str) -> str:
        """Detect Twi dialect from text content."""
        if not text or pd.isna(text):
            return 'standard'

        text_lower = str(text).lower()

        # Check for dialect markers
        for dialect, markers in self.dialect_markers.items():
            if any(marker in text_lower for marker in markers):
                return dialect

        return 'standard'

    def _detect_language(self, text: str) -> str:
        """Detect language from text content."""
        if not text or pd.isna(text):
            return 'unknown'

        text_lower = str(text).lower()

        # Twi/Akan indicators
        twi_indicators = ['ɛ', 'ɔ', 'wo', 'me', 'na', 'ne', 'ho', 'te', 'sɛn', 'da', 'ase']
        if any(indicator in text_lower for indicator in twi_indicators):
            return 'twi'

        # Ga indicators
        ga_indicators = ['ni', 'kɛ', 'lɛ', 'wɔ', 'nɛ', 'kome', 'shika']
        if any(indicator in text_lower for indicator in ga_indicators):
            return 'ga'

        # English fallback
        if any(word in text_lower for word in ['the', 'and', 'is', 'to', 'in', 'of']):
            return 'en'

        return 'twi'  # Default to Twi

    def create_unified_dataset(self) -> pd.DataFrame:
        """Create unified TTS dataset from all inventory sources."""
        print("\n🔄 Creating unified TTS dataset...")

        inventory = self.load_tts_inventory()

        all_pairs = []
        processed_datasets = 0

        for _, row in tqdm(inventory.iterrows(), total=len(inventory), desc="Processing datasets"):
            try:
                pairs = self.discover_audio_text_pairs(row)
                if pairs:
                    all_pairs.extend(pairs)
                    processed_datasets += 1
                    print(f"   Found {len(pairs)} samples in {row['data_name']}")
            except Exception as e:
                print(f"   ⚠️ Error processing {row['data_name']}: {e}")

        if all_pairs:
            unified_df = pd.DataFrame(all_pairs)

            # Add audio metadata
            unified_df = self._add_audio_metadata(unified_df)

            # Quality filtering
            unified_df = self._quality_filter(unified_df)

            # Generate speaker statistics
            self._generate_speaker_stats(unified_df)

            print(f"\n✅ Unified dataset created:")
            print(f"   Total samples: {len(unified_df)}")
            print(f"   Datasets processed: {processed_datasets}")
            print(f"   Languages: {unified_df['language'].value_counts().to_dict()}")
            print(f"   Dialects: {unified_df['dialect'].value_counts().to_dict()}")
            print(f"   Unique speakers: {unified_df['speaker_id'].nunique()}")

            return unified_df
        else:
            print("   ⚠️ No data found, creating dummy dataset for testing...")
            return self._create_dummy_dataset()

    def _add_audio_metadata(self, df: pd.DataFrame) -> pd.DataFrame:
        """Add audio metadata (duration, sample rate, etc.)."""
        print("   Adding audio metadata...")

        durations = []
        sample_rates = []
        audio_exists = []

        for audio_path in tqdm(df['audio_path'], desc="Audio metadata", leave=False):
            try:
                if Path(audio_path).exists():
                    # Get audio info without loading full file
                    info = sf.info(audio_path)
                    durations.append(info.duration)
                    sample_rates.append(info.samplerate)
                    audio_exists.append(True)
                else:
                    durations.append(0.0)
                    sample_rates.append(22050)
                    audio_exists.append(False)
            except Exception:
                durations.append(0.0)
                sample_rates.append(22050)
                audio_exists.append(False)

        df['duration'] = durations
        df['sample_rate'] = sample_rates
        df['audio_exists'] = audio_exists
        df['text_length'] = df['text'].str.len()

        return df

    def _quality_filter(self, df: pd.DataFrame) -> pd.DataFrame:
        """Apply quality filtering based on duration, text length, etc."""
        print("   Applying quality filters...")

        initial_count = len(df)

        # Filter criteria
        df = df[
            (df['audio_exists']) &
            (df['duration'] >= 0.5) &      # Minimum 0.5 seconds
            (df['duration'] <= 15.0) &     # Maximum 15 seconds for TTS
            (df['text_length'] >= 5) &     # Minimum 5 characters
            (df['text_length'] <= 500) &   # Maximum 500 characters
            (df['text'].str.strip() != '') # Non-empty text
        ].copy()

        filtered_count = len(df)
        print(f"   Filtered: {initial_count} → {filtered_count} samples ({filtered_count/initial_count*100:.1f}% retained)")

        return df

    def _generate_speaker_stats(self, df: pd.DataFrame):
        """Generate comprehensive speaker statistics."""
        print("   Generating speaker statistics...")

        speaker_stats = df.groupby('speaker_id').agg({
            'audio_path': 'count',
            'duration': ['sum', 'mean', 'std'],
            'text_length': ['mean', 'std'],
            'language': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'unknown',
            'dialect': lambda x: x.mode().iloc[0] if len(x.mode()) > 0 else 'unknown',
            'dataset': lambda x: ', '.join(x.unique())
        }).round(2)

        speaker_stats.columns = [
            'sample_count', 'total_duration', 'avg_duration', 'duration_std',
            'avg_text_length', 'text_length_std', 'language', 'dialect', 'datasets'
        ]

        self.speaker_stats = speaker_stats

        # Dataset statistics
        dataset_stats = df.groupby('dataset').agg({
            'audio_path': 'count',
            'speaker_id': 'nunique',
            'duration': 'sum',
            'language': lambda x: ', '.join(x.unique()),
            'dialect': lambda x: ', '.join(x.unique())
        }).round(2)

        dataset_stats.columns = [
            'sample_count', 'speaker_count', 'total_hours', 'languages', 'dialects'
        ]
        dataset_stats['total_hours'] = (dataset_stats['total_hours'] / 3600).round(2)

        self.dataset_stats = dataset_stats

        print(f"\n📊 Speaker Statistics:")
        print(f"   Total speakers: {len(speaker_stats)}")
        print(f"   Avg samples per speaker: {speaker_stats['sample_count'].mean():.1f}")
        print(f"   Avg duration per speaker: {speaker_stats['total_duration'].mean():.1f}s")

        print(f"\n📈 Dataset Statistics:")
        for dataset, stats in dataset_stats.head().iterrows():
            print(f"   {dataset}: {stats['sample_count']} samples, {stats['speaker_count']} speakers, {stats['total_hours']}h")

    def _create_dummy_dataset(self) -> pd.DataFrame:
        """Create dummy dataset for testing."""
        print("   Creating dummy TTS dataset...")

        # Create dummy audio directory
        dummy_dir = self.paths['tts_ready'] / 'dummy_tts'
        dummy_dir.mkdir(exist_ok=True)

        # Sample texts in different dialects
        samples = [
            ("Wo ho te sɛn? Me ho yɛ.", "twi", "standard", "speaker_001"),
            ("Me ma wo akye. Nante yie.", "twi", "asante", "speaker_002"),
            ("Me pɛ sɛ mekasa Twi yie.", "twi", "akuapem", "speaker_003"),
            ("Ɛyɛ fɛ sɛ yɛbɛhyia.", "twi", "fante", "speaker_004"),
            ("Ni kome wɔ?", "ga", "standard", "speaker_005")
        ]

        dummy_data = []
        sample_rate = 22050

        for i, (text, lang, dialect, speaker) in enumerate(samples):
            # Create dummy audio file
            audio_file = dummy_dir / f"{speaker}_{i:03d}.wav"

            if not audio_file.exists():
                # Generate 3-second dummy audio
                duration = 3.0
                dummy_audio = np.random.normal(0, 0.1, int(sample_rate * duration)).astype(np.float32)
                sf.write(audio_file, dummy_audio, sample_rate)

            dummy_data.append({
                'audio_path': str(audio_file),
                'text': text,
                'dataset': 'dummy_tts',
                'speaker_id': speaker,
                'dialect': dialect,
                'language': lang,
                'duration': duration,
                'sample_rate': sample_rate,
                'audio_exists': True,
                'text_length': len(text)
            })

        return pd.DataFrame(dummy_data)

# Load comprehensive TTS data
data_loader = ComprehensiveTTSDataLoader(PATHS)
tts_dataset = data_loader.create_unified_dataset()
```

```python
class AdvancedXTTSTrainer:
    """Advanced XTTS trainer with comprehensive features for Twi multi-speaker TTS."""

    def __init__(self, config: OptimizedXTTSConfig, dataset: pd.DataFrame):
        self.config = config
        self.dataset = dataset
        self.speaker_manager = None
        self.speaker_encoder = None
        self.xtts_model = None

        # Initialize speaker recognition for embeddings
        self._setup_speaker_recognition()

    def _setup_speaker_recognition(self):
        """Setup speaker recognition model for embeddings."""
        print("\n🎯 Setting up speaker recognition...")

        try:
            # Use SpeechBrain for speaker embeddings
            self.speaker_encoder = SpeakerRecognition.from_hparams(
                source="speechbrain/spkrec-ecapa-voxceleb",
                savedir=str(self.config.model_path / "speaker_encoder")
            )
            print("   ✅ SpeechBrain speaker encoder loaded")
        except Exception as e:
            print(f"   ⚠️ Could not load SpeechBrain: {e}")
            print("   Using simple speaker clustering...\n")
            self.speaker_encoder = None

    def generate_speaker_embeddings(self) -> Dict[str, np.ndarray]:
        """Generate speaker embeddings for all speakers in dataset."""
        print("\n🔊 Generating speaker embeddings...")

        speaker_embeddings = {}

        # Group by speaker
        speaker_groups = self.dataset.groupby('speaker_id')

        for speaker_id, group in tqdm(speaker_groups, desc="Speaker embeddings"):
            try:
                # Select best quality audio for embedding
                best_audio = self._select_best_audio_for_speaker(group)

                if best_audio is not None:
                    embedding = self._extract_speaker_embedding(best_audio)
                    if embedding is not None:
                        speaker_embeddings[speaker_id] = embedding

            except Exception as e:
                print(f"   ⚠️ Error processing speaker {speaker_id}: {e}")

        print(f"   ✅ Generated embeddings for {len(speaker_embeddings)} speakers")

        # Save embeddings
        embeddings_path = self.config.output_path / "speaker_embeddings.pkl"
        with open(embeddings_path, 'wb') as f:
            pickle.dump(speaker_embeddings, f)

        print(f"   📁 Embeddings saved to {embeddings_path}")
        return speaker_embeddings

    def _select_best_audio_for_speaker(self, speaker_group: pd.DataFrame) -> Optional[str]:
        """Select best quality audio file for speaker embedding."""
        # Filter by duration (3-8 seconds is ideal for embeddings)
        suitable = speaker_group[
            (speaker_group['duration'] >= 3.0) &
            (speaker_group['duration'] <= 8.0) &
            (speaker_group['audio_exists'])
        ]

        if len(suitable) == 0:
            # Fallback to any available audio
            suitable = speaker_group[speaker_group['audio_exists']]

        if len(suitable) == 0:
            return None

        # Select audio closest to 5 seconds
        suitable['duration_score'] = abs(suitable['duration'] - 5.0)
        best = suitable.loc[suitable['duration_score'].idxmin()]

        return best['audio_path']

    def _extract_speaker_embedding(self, audio_path: str) -> Optional[np.ndarray]:
        """Extract speaker embedding from audio file."""
        try:
            if self.speaker_encoder is not None:
                # Use SpeechBrain
                signal = self.speaker_encoder.load_audio(audio_path)
                embedding = self.speaker_encoder.encode_batch(signal.unsqueeze(0))
                return embedding.squeeze().cpu().numpy()
            else:
                # Simple spectral features as fallback
                audio, sr = librosa.load(audio_path, sr=16000)

                # Extract MFCC features
                mfccs = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=13)
                # Statistical summary
                features = np.concatenate([
                    np.mean(mfccs, axis=1),
                    np.std(mfccs, axis=1),
                    np.max(mfccs, axis=1),
                    np.min(mfccs, axis=1)
                ])

                return features

        except Exception as e:
            print(f"   ⚠️ Error extracting embedding from {audio_path}: {e}")
            return None

    def prepare_coqui_dataset(self) -> Path:
        """Prepare dataset in Coqui TTS format with speaker embeddings."""
        print("\n📄 Preparing Coqui TTS dataset...")

        # Create output directory
        dataset_dir = self.config.data_path / "coqui_xtts_dataset"
        dataset_dir.mkdir(parents=True, exist_ok=True)

        # Create subdirectories
        audio_dir = dataset_dir / "wavs"
        audio_dir.mkdir(exist_ok=True)

        metadata_lines = []
        processed_count = 0

        for idx, row in tqdm(self.dataset.iterrows(), total=len(self.dataset), desc="Processing audio"):
            try:
                # Load and preprocess audio
                audio, sr = librosa.load(row['audio_path'], sr=self.config.sample_rate)

                # Quality checks
                if len(audio) < self.config.sample_rate * self.config.min_audio_len:
                    continue
                if len(audio) > self.config.sample_rate * self.config.max_audio_len:
                    audio = audio[:int(self.config.sample_rate * self.config.max_audio_len)]

                # Normalize audio
                audio = librosa.util.normalize(audio)

                # Save processed audio
                output_filename = f"audio_{processed_count:06d}.wav"
                output_path = audio_dir / output_filename
                sf.write(output_path, audio, self.config.sample_rate)

                # Clean text
                text = self._clean_text(row['text'])

                if len(text) < self.config.min_text_len or len(text) > self.config.max_text_len:
                    continue

                # Create metadata entry
                # Format: audio_file|text|speaker_name|language|emotion
                metadata_line = f"{output_filename}|{text}|{row['speaker_id']}|{row['language']}|neutral"
                metadata_lines.append(metadata_line)

                processed_count += 1

            except Exception as e:
                print(f"   ⚠️ Error processing {row['audio_path']}: {e}")
                continue

        # Save metadata
        metadata_path = dataset_dir / "metadata.csv"
        with open(metadata_path, 'w', encoding='utf-8') as f:
            f.write('\n'.join(metadata_lines))

        # Create splits
        self._create_train_val_splits(dataset_dir, metadata_lines)

        print(f"   ✅ Processed {processed_count} samples")
        print(f"   📁 Dataset saved to: {dataset_dir}")

        return dataset_dir

    def _clean_text(self, text: str) -> str:
        """Clean and normalize text for TTS."""
        if not text or pd.isna(text):
            return ""

        text = str(text).strip()

        # Remove extra whitespace
        text = ' '.join(text.split())

        # Normalize punctuation
        import re
        text = re.sub(r'[\"\\u201c\\u201d]', '\"', text)  # Normalize quotes
        text = re.sub(r'[\\u2018\\u2019]', "\'", text)      # Normalize apostrophes
        text = re.sub(r'\\u2026', '...', text)           # Normalize ellipsis
        text = re.sub(r'[\\u2013\\u2014]', '-', text)      # Normalize dashes

        # Remove multiple consecutive punctuation
        text = re.sub(r'([.!?])\\1+', r'\\1', text)

        # Ensure sentence ends with punctuation
        if text and text[-1] not in '.!?':
            text += '.'

        return text

    def _create_train_val_splits(self, dataset_dir: Path, metadata_lines: List[str]):
        """Create train/validation splits with speaker balancing."""
        print("   Creating train/validation splits...")

        # Parse metadata to get speaker info
        speaker_samples = defaultdict(list)
        for line in metadata_lines:
            parts = line.split('|')
            if len(parts) >= 3:
                speaker_id = parts[2]
                speaker_samples[speaker_id].append(line)

        train_lines = []
        val_lines = []

        # Split each speaker's data
        for speaker_id, samples in speaker_samples.items():
            # Ensure each speaker has at least one sample in validation
            if len(samples) == 1:
                train_lines.extend(samples)
            else:
                val_count = max(1, len(samples) // 10)  # 10% for validation
                val_samples = samples[:val_count]
                train_samples = samples[val_count:]

                train_lines.extend(train_samples)
                val_lines.extend(val_samples)

        # Save splits
        with open(dataset_dir / "metadata_train.csv", 'w', encoding='utf-8') as f:
            f.write('\n'.join(train_lines))

        with open(dataset_dir / "metadata_val.csv", 'w', encoding='utf-8') as f:
            f.write('\n'.join(val_lines))

        print(f"   Train samples: {len(train_lines)}")
        print(f"   Validation samples: {len(val_lines)}")
        print(f"   Speakers in train: {len(set(line.split('|')[2] for line in train_lines))}")
        print(f"   Speakers in val: {len(set(line.split('|')[2] for line in val_lines))}")

# Initialize advanced XTTS trainer
if len(tts_dataset) > 0:
    xtts_trainer = AdvancedXTTSTrainer(xtts_config, tts_dataset)
    print("\n✅ Advanced XTTS trainer initialized!")
else:
    print("\n⚠️ No TTS data available for training")
    xtts_trainer = None
```

```python
class AdvancedXTTSTrainer:
    """Advanced XTTS trainer with comprehensive features for Twi multi-speaker TTS."""

    def __init__(self, config: OptimizedXTTSConfig, dataset: pd.DataFrame):
        self.config = config
        self.dataset = dataset
        self.speaker_manager = None
        self.speaker_encoder = None
        self.xtts_model = None

        # Initialize speaker recognition for embeddings
        self._setup_speaker_recognition()

    def _setup_speaker_recognition(self):
        """Setup speaker recognition model for embeddings."""
        print("\n🎯 Setting up speaker recognition...")

        try:
            # Use SpeechBrain for speaker embeddings
            self.speaker_encoder = SpeakerRecognition.from_hparams(
                source="speechbrain/spkrec-ecapa-voxceleb",
                savedir=str(self.config.model_path / "speaker_encoder")
            )
            print("   ✅ SpeechBrain speaker encoder loaded")
        except Exception as e:
            print(f"   ⚠️ Could not load SpeechBrain: {e}")
            print("   Using simple speaker clustering...\n")
            self.speaker_encoder = None

    def generate_speaker_embeddings(self) -> Dict[str, np.ndarray]:
        """Generate speaker embeddings for all speakers in dataset."""
        print("\n🔊 Generating speaker embeddings...")

        speaker_embeddings = {}

        # Group by speaker
        speaker_groups = self.dataset.groupby('speaker_id')

        for speaker_id, group in tqdm(speaker_groups, desc="Speaker embeddings"):
            try:
                # Select best quality audio for embedding
                best_audio = self._select_best_audio_for_speaker(group)

                if best_audio is not None:
                    embedding = self._extract_speaker_embedding(best_audio)
                    if embedding is not None:
                        speaker_embeddings[speaker_id] = embedding

            except Exception as e:
                print(f"   ⚠️ Error processing speaker {speaker_id}: {e}")

        print(f"   ✅ Generated embeddings for {len(speaker_embeddings)} speakers")

        # Save embeddings
        embeddings_path = self.config.output_path / "speaker_embeddings.pkl"
        with open(embeddings_path, 'wb') as f:
            pickle.dump(speaker_embeddings, f)

        print(f"   📁 Embeddings saved to {embeddings_path}")
        return speaker_embeddings

    def _select_best_audio_for_speaker(self, speaker_group: pd.DataFrame) -> Optional[str]:
        """Select best quality audio file for speaker embedding."""
        # Filter by duration (3-8 seconds is ideal for embeddings)
        suitable = speaker_group[
            (speaker_group['duration'] >= 3.0) &
            (speaker_group['duration'] <= 8.0) &
            (speaker_group['audio_exists'])
        ]

        if len(suitable) == 0:
            # Fallback to any available audio
            suitable = speaker_group[speaker_group['audio_exists']]

        if len(suitable) == 0:
            return None

        # Select audio closest to 5 seconds
        suitable['duration_score'] = abs(suitable['duration'] - 5.0)
        best = suitable.loc[suitable['duration_score'].idxmin()]

        return best['audio_path']

    def _extract_speaker_embedding(self, audio_path: str) -> Optional[np.ndarray]:
        """Extract speaker embedding from audio file."""
        try:
            if self.speaker_encoder is not None:
                # Use SpeechBrain
                signal = self.speaker_encoder.load_audio(audio_path)
                embedding = self.speaker_encoder.encode_batch(signal.unsqueeze(0))
                return embedding.squeeze().cpu().numpy()
            else:
                # Simple spectral features as fallback
                audio, sr = librosa.load(audio_path, sr=16000)

                # Extract MFCC features
                mfccs = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=13)
                # Statistical summary
                features = np.concatenate([
                    np.mean(mfccs, axis=1),
                    np.std(mfccs, axis=1),
                    np.max(mfccs, axis=1),
                    np.min(mfccs, axis=1)
                ])

                return features

        except Exception as e:
            print(f"   ⚠️ Error extracting embedding from {audio_path}: {e}")
            return None

    def prepare_coqui_dataset(self) -> Path:
        """Prepare dataset in Coqui TTS format with speaker embeddings."""
        print("\n📄 Preparing Coqui TTS dataset...")

        # Create output directory
        dataset_dir = self.config.data_path / "coqui_xtts_dataset"
        dataset_dir.mkdir(parents=True, exist_ok=True)

        # Create subdirectories
        audio_dir = dataset_dir / "wavs"
        audio_dir.mkdir(exist_ok=True)

        metadata_lines = []
        processed_count = 0

        for idx, row in tqdm(self.dataset.iterrows(), total=len(self.dataset), desc="Processing audio"):
            try:
                # Load and preprocess audio
                audio, sr = librosa.load(row['audio_path'], sr=self.config.sample_rate)

                # Quality checks
                if len(audio) < self.config.sample_rate * self.config.min_audio_len:
                    continue
                if len(audio) > self.config.sample_rate * self.config.max_audio_len:
                    audio = audio[:int(self.config.sample_rate * self.config.max_audio_len)]

                # Normalize audio
                audio = librosa.util.normalize(audio)

                # Save processed audio
                output_filename = f"audio_{processed_count:06d}.wav"
                output_path = audio_dir / output_filename
                sf.write(output_path, audio, self.config.sample_rate)

                # Clean text
                text = self._clean_text(row['text'])

                if len(text) < self.config.min_text_len or len(text) > self.config.max_text_len:
                    continue

                # Create metadata entry
                # Format: audio_file|text|speaker_name|language|emotion
                metadata_line = f"{output_filename}|{text}|{row['speaker_id']}|{row['language']}|neutral"
                metadata_lines.append(metadata_line)

                processed_count += 1

            except Exception as e:
                print(f"   ⚠️ Error processing {row['audio_path']}: {e}")
                continue

        # Save metadata
        metadata_path = dataset_dir / "metadata.csv"
        with open(metadata_path, 'w', encoding='utf-8') as f:
            f.write('\n'.join(metadata_lines))

        # Create splits
        self._create_train_val_splits(dataset_dir, metadata_lines)

        print(f"   ✅ Processed {processed_count} samples")
        print(f"   📁 Dataset saved to: {dataset_dir}")

        return dataset_dir

    def _clean_text(self, text: str) -> str:
        """Clean and normalize text for TTS."""
        if not text or pd.isna(text):
            return ""

        text = str(text).strip()

        # Remove extra whitespace
        text = ' '.join(text.split())

        # Normalize punctuation
        import re
        text = re.sub(r'[\"\\u201c\\u201d]', '\"', text)  # Normalize quotes
        text = re.sub(r'[\\u2018\\u2019]', "\'", text)      # Normalize apostrophes
        text = re.sub(r'\\u2026', '...', text)           # Normalize ellipsis
        text = re.sub(r'[\\u2013\\u2014]', '-', text)      # Normalize dashes

        # Remove multiple consecutive punctuation
        text = re.sub(r'([.!?])\\1+', r'\\1', text)

        # Ensure sentence ends with punctuation
        if text and text[-1] not in '.!?':
            text += '.'

        return text

    def _create_train_val_splits(self, dataset_dir: Path, metadata_lines: List[str]):
        """Create train/validation splits with speaker balancing."""
        print("   Creating train/validation splits...")

        # Parse metadata to get speaker info
        speaker_samples = defaultdict(list)
        for line in metadata_lines:
            parts = line.split('|')
            if len(parts) >= 3:
                speaker_id = parts[2]
                speaker_samples[speaker_id].append(line)

        train_lines = []
        val_lines = []

        # Split each speaker's data
        for speaker_id, samples in speaker_samples.items():
            # Ensure each speaker has at least one sample in validation
            if len(samples) == 1:
                train_lines.extend(samples)
            else:
                val_count = max(1, len(samples) // 10)  # 10% for validation
                val_samples = samples[:val_count]
                train_samples = samples[val_count:]

                train_lines.extend(train_samples)
                val_lines.extend(val_samples)

        # Save splits
        with open(dataset_dir / "metadata_train.csv", 'w', encoding='utf-8') as f:
            f.write('\n'.join(train_lines))

        with open(dataset_dir / "metadata_val.csv", 'w', encoding='utf-8') as f:
            f.write('\n'.join(val_lines))

        print(f"   Train samples: {len(train_lines)}")
        print(f"   Validation samples: {len(val_lines)}")
        print(f"   Speakers in train: {len(set(line.split('|')[2] for line in train_lines))}")
        print(f"   Speakers in val: {len(set(line.split('|')[2] for line in val_lines))}")

# Initialize advanced XTTS trainer
if len(tts_dataset) > 0:
    xtts_trainer = AdvancedXTTSTrainer(xtts_config, tts_dataset)
    print("\n✅ Advanced XTTS trainer initialized!")
else:
    print("\n⚠️ No TTS data available for training")
    xtts_trainer = None
```

```python
class OptimizedXTTSInference:
    """Optimized XTTS inference engine for voice cloning and multi-speaker TTS."""

    def __init__(self, config: OptimizedXTTSConfig):
        self.config = config
        self.model = None
        self.speaker_embeddings = {}
        self.available_models = []

        # Load available models
        self._discover_available_models()

    def _discover_available_models(self):
        """Discover available XTTS models."""
        print("🔍 Discovering available XTTS models...")

        if TTS_AVAILABLE:
            try:
                # List available TTS models
                tts = TTS()
                self.available_models = [model for model in tts.list_models() if 'xtts' in model]
                print(f"   Found {len(self.available_models)} XTTS models")
                for model in self.available_models:
                    print(f"   - {model}")
            except Exception as e:
                print(f"   ⚠️ Error discovering models: {e}")

        # Add default models
        default_models = [
            "tts_models/multilingual/multi-dataset/xtts_v2",
            "tts_models/multilingual/multi-dataset/xtts_v1.1"
        ]

        for model in default_models:
            if model not in self.available_models:
                self.available_models.append(model)

        print(f"   Total available models: {len(self.available_models)}")

    def load_model(self, model_name: str = "tts_models/multilingual/multi-dataset/xtts_v2"):
        """Load XTTS model for inference."""
        print(f"\n🤖 Loading XTTS model: {model_name}")

        try:
            if TTS_AVAILABLE:
                self.model = TTS(model_name, gpu=torch.cuda.is_available())
                print(f"   ✅ Model loaded successfully")

                # Load speaker embeddings if available
                embeddings_path = self.config.output_path / "speaker_embeddings.pkl"
                if embeddings_path.exists():
                    with open(embeddings_path, 'rb') as f:
                        self.speaker_embeddings = pickle.load(f)
                    print(f"   📁 Loaded {len(self.speaker_embeddings)} speaker embeddings")

            else:
                print("   ⚠️ TTS not available, using mock model")
                self.model = self._create_mock_model()

        except Exception as e:
            print(f"   ❌ Error loading model: {e}")
            self.model = None

    def _create_mock_model(self):
        """Create mock model for testing when TTS is not available."""
        class MockXTTSModel:
            def tts_to_file(self, text, speaker_wav=None, language="en", file_path="output.wav"):
                # Generate dummy audio
                dummy_audio = np.random.normal(0, 0.1, 22050 * 3).astype(np.float32)
                sf.write(file_path, dummy_audio, 22050)
                return file_path

            def tts(self, text, speaker_wav=None, language="en"):
                return np.random.normal(0, 0.1, 22050 * 3).astype(np.float32)

        return MockXTTSModel()

    def clone_voice(self, text: str, reference_audio: str, language: str = "en",
                   output_path: Optional[str] = None) -> str:
        """Clone voice using reference audio."""
        if not self.model:
            raise ValueError("Model not loaded. Call load_model() first.")

        if output_path is None:
            output_path = self.config.output_path / f"cloned_{int(time.time())}.wav"
        else:
            output_path = Path(output_path)

        print(f"🎭 Cloning voice for text: '{text[:50]}...'\n")
        print(f"   Reference: {Path(reference_audio).name}\n")
        print(f"   Language: {language}\n")

        try:
            # Perform voice cloning
            self.model.tts_to_file(
                text=text,
                speaker_wav=reference_audio,
                language=language,
                file_path=str(output_path)
            )

            print(f"   ✅ Voice cloned to: {output_path}\n")
            return str(output_path)

        except Exception as e:
            print(f"   ❌ Voice cloning failed: {e}\n")
            raise

    def multi_speaker_synthesis(self, texts_and_speakers: List[Tuple[str, str]],
                               language: str = "en", output_dir: Optional[Path] = None) -> List[str]:
        """Synthesize multiple texts with different speakers."""
        if output_dir is None:
            output_dir = self.config.output_path / "multi_speaker_output"

        output_dir.mkdir(parents=True, exist_ok=True)

        results = []

        for i, (text, speaker_ref) in enumerate(texts_and_speakers):
            try:
                output_path = output_dir / f"speaker_{i:03d}.wav"

                cloned_path = self.clone_voice(
                    text=text,
                    reference_audio=speaker_ref,
                    language=language,
                    output_path=str(output_path)
                )

                results.append(cloned_path)

            except Exception as e:
                print(f"   ⚠️ Failed to synthesize for speaker {i}: {e}\n")
                results.append(None)

        success_count = sum(1 for r in results if r is not None)
        print(f"\n✅ Multi-speaker synthesis complete: {success_count}/{len(texts_and_speakers)} successful\n")

        return results

    def test_voice_cloning(self, test_phrases: Optional[List[str]] = None):
        """Test voice cloning with sample phrases."""
        if test_phrases is None:
            test_phrases = [
                "Wo ho te sɛn? Me ho yɛ.",
                "Me ma wo akye. Ɛyɛ sɛn?",
                "Me pɛ sɛ mekasa Twi yie.",
                "Medaase paa! Nante yie.",
                "This is a test of English voice cloning."
            ]

        print("\n🧪 Testing voice cloning capabilities...\n")

        # Find reference audio files
        reference_audios = []
        if len(tts_dataset) > 0:
            # Use actual dataset samples
            sample_speakers = tts_dataset['speaker_id'].unique()[:3]
            for speaker in sample_speakers:
                speaker_data = tts_dataset[tts_dataset['speaker_id'] == speaker]
                if len(speaker_data) > 0:
                    reference_audios.append(speaker_data.iloc[0]['audio_path'])

        if not reference_audios:
            # Create dummy reference audio
            print("   Creating dummy reference audio for testing...\n")
            dummy_ref = self.config.output_path / "dummy_reference.wav"
            dummy_audio = np.random.normal(0, 0.1, 22050 * 5).astype(np.float32)
            sf.write(dummy_ref, dummy_audio, 22050)
            reference_audios = [str(dummy_ref)]

        test_results = []

        for i, phrase in enumerate(test_phrases):
            try:
                ref_audio = reference_audios[i % len(reference_audios)]
                language = "tw" if any(char in phrase for char in "ɛɔ") else "en"

                output_path = self.clone_voice(
                    text=phrase,
                    reference_audio=ref_audio,
                    language=language
                )

                test_results.append({
                    "text": phrase,
                    "reference": Path(ref_audio).name,
                    "output": output_path,
                    "language": language,
                    "status": "success"
                })

            except Exception as e:
                test_results.append({
                    "text": phrase,
                    "status": "failed",
                    "error": str(e)
                })

        # Save test results
        results_path = self.config.output_path / "voice_cloning_test_results.json"
        with open(results_path, 'w', encoding='utf-8') as f:
            json.dump(test_results, f, indent=2, ensure_ascii=False)

        success_count = sum(1 for r in test_results if r["status"] == "success")
        print(f"\n📊 Test Results: {success_count}/{len(test_phrases)} successful\n")
        print(f"📁 Results saved to: {results_path}\n")

        return test_results

    def benchmark_performance(self, num_samples: int = 10):
        """Benchmark voice cloning performance."""
        print(f"\n📊 Benchmarking performance with {num_samples} samples...\n")

        if not self.model:
            print("   ⚠️ Model not loaded\n")
            return

        # Sample texts of different lengths
        short_text = "Hello world."
        medium_text = "This is a longer sentence to test medium length synthesis."
        long_text = "This is a very long sentence that will test the model's ability to synthesize longer passages of text with consistent quality and naturalness."

        test_texts = [short_text, medium_text, long_text]

        # Get reference audio
        if len(tts_dataset) > 0:
            ref_audio = tts_dataset.iloc[0]['audio_path']
        else:
            # Create dummy reference
            ref_audio = self.config.output_path / "benchmark_ref.wav"
            dummy_audio = np.random.normal(0, 0.1, 22050 * 3).astype(np.float32)
            sf.write(ref_audio, dummy_audio, 22050)
            ref_audio = str(ref_audio)

        results = []

        for text_type, text in zip(["short", "medium", "long"], test_texts):
            times = []

            for i in range(num_samples):
                start_time = time.time()

                try:
                    output_path = self.config.output_path / f"benchmark_{text_type}_{i}.wav"
                    self.clone_voice(
                        text=text,
                        reference_audio=ref_audio,
                        output_path=str(output_path)
                    )

                    elapsed = time.time() - start_time
                    times.append(elapsed)

                except Exception as e:
                    print(f"   ⚠️ Benchmark failed for {text_type} sample {i}: {e}\n")
            if times:
                avg_time = np.mean(times)
                std_time = np.std(times)

                results.append({
                    "text_type": text_type,
                    "text_length": len(text),
                    "avg_time": avg_time,
                    "std_time": std_time,
                    "min_time": min(times),
                    "max_time": max(times),
                    "samples": len(times)
                })

                print(f"   {text_type.capitalize()}: {avg_time:.2f}s ± {std_time:.2f}s\n")

        # Save benchmark results
        benchmark_path = self.config.output_path / "benchmark_results.json"
        with open(benchmark_path, 'w') as f:
            json.dump(results, f, indent=2)

        print(f"\n📁 Benchmark results saved to: {benchmark_path}\n")
        return results

# Initialize XTTS inference engine
xtts_inference = OptimizedXTTSInference(xtts_config)
print("\n✅ XTTS inference engine initialized!\n")
```
