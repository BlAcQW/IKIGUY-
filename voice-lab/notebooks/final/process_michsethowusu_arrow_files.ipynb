```python
# %% [markdown]
# # Final Test for MichSetHowusu Arrow Datasets (Working Version)
#
# This notebook confirms the integrity of the `michsethowusu` audio datasets by correctly handling the streaming `AudioDecoder` object.

# %% [code]
# Step 1: Install and import necessary libraries
print("📦 Installing required libraries...")
!pip install -q datasets IPython torchcodec

from datasets import load_dataset
from IPython.display import display, Audio
import os

# %% [code]
# Step 2: Configure Hugging Face Token
from google.colab import userdata

try:
    HF_TOKEN = userdata.get("HF_TOKEN")
    os.environ["HF_TOKEN"] = HF_TOKEN
    print("✅ Hugging Face token configured successfully.")
except Exception as e:
    print("⚠️ HF_TOKEN not found in Colab secrets.")
    HF_TOKEN = None

# %% [code]
# Step 3: Stream and Decode a Valid Sample
dataset_name = "michsethowusu/twi_multispeaker_audio_transcribed"

print(f"--- Final Test on: {dataset_name} ---")

try:
    print("\n🔄 Connecting to the dataset stream...")
    dataset = load_dataset(dataset_name, split="train", streaming=True, token=HF_TOKEN)
    print("✅ Successfully connected.")

    print("\n📥 Searching for a valid sample...")
    valid_sample = None
    search_limit = 20

    for i, sample in enumerate(dataset):
        # *** THE FIX IS HERE: We check for the presence of the objects themselves ***
        transcription = sample.get('transcription')
        audio_object = sample.get('audio_path')

        # The validity check is now simpler: just see if the keys exist and are not None.
        if transcription and audio_object:
            print(f"✅ Found a valid sample at index {i}!")
            valid_sample = sample
            break

        if i >= search_limit - 1:
            print(f"❌ Searched the first {search_limit} samples and could not find a valid one.")
            break

    # --- Verification ---
    if valid_sample:
        print("\n--- Sample Verification ---")

        # 1. Display the text transcript
        transcription = valid_sample['transcription']
        print(f"📜 Transcription: {transcription}")

        # 2. Decode the audio object and play the sound
        # Accessing the key 'audio_path' now triggers the decoding
        decoded_audio = valid_sample['audio_path']
        audio_data = decoded_audio['array']
        sampling_rate = decoded_audio['sampling_rate']

        print(f"🎧 Audio Sample Rate: {sampling_rate} Hz")
        print("   Playing audio now...")
        display(Audio(data=audio_data, rate=sampling_rate))
    else:
        print("\nCould not verify the dataset with a valid sample.")

except Exception as e:
    print(f"\n❌ An error occurred during the test: {e}")
```
